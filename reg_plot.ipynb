```python
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
tips=sns.load_dataset('tips')
tips.head()
```

```python
sns.regplot(data=tips,x='total_bill',y='tip')
```

```python
sns.regplot(data=tips,x='total_bill',y='tip',color='red',marker='+')
```

```python
sns.regplot(data=tips,x='total_bill',y='tip',color='red',marker='+',line_kws=dict(color='k'))
```

```python
sns.regplot(data=tips,x='total_bill',y='tip',color='red',marker='+',line_kws=dict(color='k',linestyle='--'))
```

```python
sns.regplot(data=tips,x='total_bill',y='tip',color='red',line_kws=dict(color='k',linestyle='--'),
    scatter_kws=dict(s=100,color='green',alpha=0.5,edgecolor='k'))
```

```python
sns.regplot(data=tips,x='total_bill',y='tip',color='red',line_kws=dict(color='k',linestyle='--'),
    scatter_kws=dict(s=100,color='green',alpha=0.5,edgecolor='k'),ci=60)
```
