```python
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
df=pd.read_excel('sales.xlsx')
df.head()
```

```python
sns.violinplot(y='Total',data=df)
```

```python
sns.violinplot(y='Total',x='Payment',data=df)
```

```python
sns.violinplot(y='Total',x='Payment',hue='Gender',data=df)
```

```python
sns.violinplot(y='Total',x='Payment',hue='Gender',data=df,split=True)
```

```python
sns.violinplot(y='Total',x='Payment',hue='Gender',data=df,split=True,inner='stick',bw=0.2)
```

```python
sns.violinplot(y='Total',x='Payment',hue='Gender',data=df,split=True,inner='quartille',bw=.2)
```

```python
sns.violinplot(y='Total',x='Payment',hue='Gender',data=df,split=True,inner='quartille',bw=0.2,cut=0)
```
