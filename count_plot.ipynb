```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
import numpy as np
```

```python
url=('sales.csv')
```

```python
df=pd.read_csv(url)
```

```python
df.head()
```

```python
plt.figure(figsize=(15,5))
sns.countplot(x='product_name',hue='gender',data=df)
```

```python
plt.figure(figsize=(15,5))
sns.countplot(x='product_name',hue='gender',palette='PRGn',data=df)
```

```python
plt.figure(figsize=(15,5))
sns.countplot(x='product_name',hue='gender',data=df,facecolor=(1,1,1,0),linewidth=5,edgecolor='black')
```
