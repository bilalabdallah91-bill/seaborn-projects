```python
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
df=pd.read_excel('sales.xlsx')
df.head()
```

```python
sns.catplot(data=df,y='Total',x='Payment',palette='Set2')
```

```python
sns.catplot(data=df,y='Total',x='Payment',kind='box',palette='Set2')
```

```python
sns.catplot(data=df,y='Total',x='Payment',palette='Set2',kind='violin')
```

```python
sns.catplot(data=df,y='Total',x='Payment',palette='Set2',col='Gender',kind='violin')
```

```python
sns.catplot(data=df,y='Total',x='Payment',palette='Set2',col='Gender',kind='violin',row='Customer type')
```

```python
sns.catplot(data=df,y='Total',x='Payment',col='Gender',kind='violin',row='Customer type',color='gray')
```

```python
sns.catplot(data=df,y='Total',x='Payment',col='Gender',kind='violin',row='Customer type',palette='CMRmap_r')
```
