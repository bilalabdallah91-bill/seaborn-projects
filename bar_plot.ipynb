```python
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as pyplot
```

```python
mart=pd.read_csv('sales.csv')
```

```python
mart.head()
```

```python
sns.barplot(x='product_name',y='unit_price',data=mart)
```

```python
sns.barplot(x='product_name',y='unit_price',hue='gender',data=mart)
```

```python
sns.barplot(y='product_name',x='unit_price',hue='gender',data=mart)
```

```python
sns.barplot(y='product_name',x='unit_price',hue='gender',data=mart,errorbar=None)
```

```python
sns.barplot(x='product_name',y='unit_price',hue='gender',order=['Orange Juice','Apple','Detergent','Notebook','Shampoo'],hue_order=['Female','Male'],data=mart,estimator=np.sum)
```

```python
sns.barplot(x='product_name',y='unit_price',hue='gender',order=['Orange Juice','Apple','Detergent','Notebook','Shampoo'],
hue_order=['Female','Male'],data=mart,estimator=np.sum,color='black')
```

```python
sns.barplot(x='product_name',y='unit_price',order=['Orange Juice','Apple','Detergent','Notebook','Shampoo'],
hue_order=['Female','Male'],data=mart,estimator=np.sum,color='red')
```

```python
sns.barplot(x='product_name',y='unit_price',order=['Orange Juice','Apple','Detergent','Notebook','Shampoo'],
hue_order=['Female','Male'],data=mart,estimator=np.sum,palette='cividis')
```

```python
sns.barplot(x='product_name',y='unit_price',hue='gender',order=['Orange Juice','Apple','Detergent','Notebook','Shampoo'],
hue_order=['Female','Male'],data=mart,estimator=np.sum,palette='cividis',saturation=0.1)
```

```python
sns.barplot(x='product_name',y='unit_price',hue='gender',order=['Orange Juice','Apple','Detergent','Notebook','Shampoo'],
hue_order=['Female','Male'],data=mart,palette='cividis',saturation=0.1,estimator=np.median)
```
