```python
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
df=pd.read_excel('sales.xlsx')
df.head()
```

```python
sns.swarmplot(data=df,y='Total')
```

```python
sns.swarmplot(data=df,y='Total',x='Payment')
```

```python
sns.swarmplot(data=df,y='Total',x='Payment',hue='Gender')
```

```python
sns.swarmplot(data=df,y='Total',x='Payment',hue='Gender',dodge=True)
```

```python
plt.figure(figsize=(15,5))
sns.swarmplot(data=df,y='Total',x='Payment',marker='*',size=10)
```

```python
plt.figure(figsize=(15,5))
sns.swarmplot(data=df,y='Total',x='Payment',marker='*',size=10,color='black')
```

```python
plt.figure(figsize=(15,5))
sns.swarmplot(data=df,y='Total',x='Payment',marker='*',size=10,color='green')
```

```python
plt.figure(figsize=(15,5))
sns.swarmplot(data=df,y='Total',x='Payment',size=10,color='green',edgecolor='black')
```

```python
sns.boxplot(data=df,y='Total',x='Payment')
sns.swarmplot(data=df,y='Total',x='Payment')
```

```python
sns.boxplot(data=df,y='Total',x='Payment')
sns.swarmplot(data=df,y='Total',x='Payment',color='black')
```

```python
sns.violinplot(data=df,y='Total',x='Payment')
sns.swarmplot(data=df,y='Total',x='Payment')
```

```python
sns.violinplot(data=df,y='Total',x='Payment')
sns.swarmplot(data=df,y='Total',x='Payment',color='black')
```

```python
sns.violinplot(data=df,y='Total',x='Payment',inner=None)
sns.swarmplot(data=df,y='Total',x='Payment',color='white')
```

```python
sns.violinplot(data=df,y='Total',x='Payment',inner=None,width=1)
sns.swarmplot(data=df,y='Total',x='Payment',color='white')
```
