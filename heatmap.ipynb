```python
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
mart=pd.read_excel('mart_lineplot.xlsx')
mart.columns=mart.columns.str.lower()
mart.head(10)
```

```python
mart_piv=mart.pivot_table(index='outlet_year',columns='outlet_size',values='sales')
mart_piv
```

```python
sns.heatmap(mart_piv)
```

```python
sns.set_style('white')
plt.Figure(figsize=(5,5))
sns.heatmap(mart_piv)
```

```python
sns.set_style('white')
plt.Figure(figsize=(5,5))
sns.heatmap(mart_piv,annot=True)
```

```python
sns.set_style('white')
plt.Figure(figsize=(5,5))
sns.heatmap(mart_piv,annot=True,fmt='.0f')
```

```python
sns.set_style('white')
plt.Figure(figsize=(5,5))
sns.heatmap(mart_piv,annot=True,fmt='.0f',annot_kws=dict(size=15,weight='bold'))
```

```python
sns.set_style('white')
plt.Figure(figsize=(5,5))
sns.heatmap(mart_piv,annot=True,fmt='.0f',annot_kws=dict(size=15,weight='bold'),linewidth=0.5,cmap='OrRd_r')
```

```python
mart.corr(numeric_only=True)
```

```python
mart_corr=mart.corr(numeric_only=True)
sns.heatmap(mart_corr)
```

```python
mart_corr=mart.corr(numeric_only=True)
sns.heatmap(mart_corr,vmin=-1,vmax=1,center=0)
```

```python
sns.heatmap(mart_corr,vmin=-1,vmax=1,center=0,cmap='OrRd_r',annot=True,fmt='.1f'
            ,annot_kws=dict(size=15,weight='bold'),linewidth=.5,linecolor='black')
```
