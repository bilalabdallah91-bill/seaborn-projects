```python
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
data=pd.read_excel('mart_lineplot.xlsx').sample(n=50)
data.columns=data.columns.str.lower()
data.head()
```

```python
sns.relplot(data=data)
```

```python
sns.relplot(data=data,x='outlet_year',y='sales')
```

```python
sns.relplot(data=data,x='outlet_year',y='sales',kind='line')
```

```python
sns.relplot(data=data,x='outlet_year',y='sales',kind='line',ci=None)
```

```python
sns.relplot(data=data,x='outlet_year',y='sales',s=100)
```

```python
sns.relplot(data=data,x='outlet_year',y='sales',s=100,hue='outlet_size')
```

```python
sns.relplot(data=data,x='outlet_year',y='sales',s=100,hue='outlet_size',col='tier')
```

```python
sns.relplot(data=data,x='outlet_year',y='sales',s=100,hue='outlet_size',col='item_type',col_wrap=3)
```

```python
sns.relplot(data=data,x='outlet_year',y='sales',s=100,hue='outlet_size',col='outlet_size',row='tier')
```

```python
sns.relplot(data=data,x='outlet_year',y='sales',s=100,hue='outlet_size',palette=('yellow','red','green'))
```
