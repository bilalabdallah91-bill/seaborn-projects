```python
i=sns.jointplot(data=iris,x='sepal_length',y='petal_length',color='red')
i.plot_joint(sns.kdeplot)
```

```python
i=sns.jointplot(data=iris,x='sepal_length',y='petal_length',joint_kws=dict(color='red'))
i.plot_joint(sns.kdeplot)
i.plot_joint(sns.rugplot)
```

```python
i=sns.jointplot(data=iris,x='sepal_length',y='petal_length',joint_kws=dict(color='red'))
i.plot_joint(sns.kdeplot)
i.plot_joint(sns.rugplot,height=0.1,color='red')
```
