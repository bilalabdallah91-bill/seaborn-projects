```python
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt
import statistics as sts
```

```python
df=pd.read_excel('sales.xlsx')
df=df[['Gender','Payment','Unit price','Quantity','Total','gross income']]
df.head(10)
```

```python
plt.figure(figsize=(15,5))
sns.rugplot(data=df,x='Unit price')
```

```python
plt.figure(figsize=(15,5))
sns.rugplot(data=df,x='Unit price',y='gross income')
```

```python
plt.figure(figsize=(15,5))
sns.rugplot(data=df,x='Unit price',y='gross income',hue='Gender')
```

```python
plt.figure(figsize=(15,5))
sns.rugplot(data=df,x='Unit price',y='gross income',hue='Gender',height=0.05)
```

```python
plt.figure(figsize=(15,5))
sns.kdeplot(data=df,x='Unit price',y='gross income',hue='Gender')
sns.rugplot(data=df,x='Unit price',y='gross income',hue='Gender')
```

```python
plt.figure(figsize=(15,5))
sns.kdeplot(data=df,x='Unit price',y='gross income',hue='Gender',fill=True)
sns.rugplot(data=df,x='Unit price',y='gross income',hue='Gender')
```

```python
plt.figure(figsize=(15,5))
sns.scatterplot(data=df,x='Unit price',y='gross income',hue='Gender')
sns.rugplot(data=df,x='Unit price',y='gross income',hue='Gender')
```

```python
plt.figure(figsize=(15,5))
sns.kdeplot(data=df,x='Unit price',y='gross income',hue='Gender',fill=True)
sns.rugplot(data=df,x='Unit price',y='gross income',hue='Gender',height=-0.01,clip_on=False)
```
