```python
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as pyplot
```

```python
df=pd.read_excel('sales.xlsx')
df.head()
```

```python
sns.stripplot(data=df,y='Total')
```

```python
sns.stripplot(data=df,y='Total',jitter=0.2)
```

```python
sns.stripplot(data=df,y='Total',jitter=0.2,linewidth=.2)
```

```python
sns.stripplot(data=df,y='Total',x='Payment',hue='Payment',jitter=.2,linewidth=.5)
```

```python
sns.stripplot(data=df,y='Total',x='Payment',hue='Gender',jitter=.2,linewidth=.5)
```

```python
sns.stripplot(data=df,y='Total',x='Payment',hue='Gender',jitter=.2,linewidth=.5,dodge=True)
```

```python
sns.stripplot(data=df,y='Total',x='Payment',jitter=.1)
sns.violinplot(data=df,y='Total',x='Payment')
```

```python
sns.stripplot(data=df,y='Total',x='Payment',jitter=.1)
sns.boxplot(data=df,y='Total',x='Payment')
```

```python
sns.stripplot(data=df,y='Total',x='Payment',jitter=.1,color='k')
sns.boxplot(data=df,y='Total',x='Payment')
```
