```python
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt
import statistics as sts
pd.set_option('display.max_columns',None)
```

```python
iris=sns.load_dataset('iris')
iris.head()
```

```python
sns.pairplot(data=iris)
```

```python
sns.pairplot(data=iris,diag_kind='kde')
```

```python
sns.pairplot(data=iris,kind='reg')
```

```python
sns.pairplot(data=iris,hue='species')
```

```python
sns.pairplot(data=iris,x_vars=['sepal_width','petal_width','petal_length'],y_vars='sepal_length')
```

```python
sns.pairplot(data=iris,x_vars=['sepal_width','petal_width','petal_length'],y_vars=['sepal_length','sepal_width'],diag_kind=None)
```

```python
sns.pairplot(data=iris,corner=True)
```

```python
sns.pairplot(data=iris,diag_kws=dict(color='green',kde=True))
```

```python
sns.pairplot(data=iris,diag_kws=dict(color='green',kde=True),plot_kws=dict(color='red'))
```

```python
sns.pairplot(data=iris,diag_kws=dict(color='green',kde=True),plot_kws=dict(color='red',marker=10,s=100))
```

```python
i=sns.pairplot(data=iris,plot_kws=dict(color='red'))
i.map_upper(sns.kdeplot)
```

```python
i=sns.pairplot(data=iris,plot_kws=dict(color='red'))
i.map_upper(sns.kdeplot)
i.map_lower(sns.kdeplot)
```
